## 🅐 Section A — Bivariate QQ Granger Causality: `qq_causality()`

Tests whether the **τ-quantile of x_{t-1}** Granger-causes the **θ-quantile of y_t**.  
Returns a full (θ, τ) grid of slope estimates and t-statistics.

```r
qq_causality(x, y,
             y_quantiles = seq(0.05, 0.95, by = 0.05),
             x_quantiles = seq(0.05, 0.95, by = 0.05),
             bandwidth = 0.05, n_boot = 200,
             cdf_based_kernel = TRUE,
             cause_name = "X", effect_name = "Y",
             verbose = TRUE, seed = 42)
```

### A.1 — Simulate Toy Data (Bivariate)

````python
%%R
set.seed(42)
n <- 200

x <- rnorm(n)                                 # cause variable  (e.g. oil price returns)
y <- 0.3 * c(0, x[-n]) + rnorm(n, sd = 0.4)  # effect variable (e.g. stock returns)

cat("x: n =", length(x), " | mean =", round(mean(x), 3), "\n")
cat("y: n =", length(y), " | mean =", round(mean(y), 3), "\n")
````

### A.2 — Run `qq_causality()`

````python
%%R
fit_bivariate <- qq_causality(
  x             = x,              # REQUIRED ► your cause variable vector
  y             = y,              # REQUIRED ► your effect variable vector

  y_quantiles   = seq(0.1, 0.9, by = 0.1),  # theta grid — quantiles of y
  x_quantiles   = seq(0.1, 0.9, by = 0.1),  # tau   grid — quantiles of x

  bandwidth        = 0.05,   # Gaussian kernel bandwidth on the [0,1] CDF scale
  n_boot           = 200,    # bootstrap replicates for p-values (use 500+ for papers)
  cdf_based_kernel = TRUE,   # TRUE = rank-based CDF kernel (recommended)
  cause_name       = "Oil_Price",    # label for x in output and plots
  effect_name      = "Stock_Return", # label for y in output and plots
  verbose          = TRUE,   # show progress bar?
  seed             = 42      # RNG seed for reproducibility
)
````

### A.3 — Print Bivariate Results

````python
%%R
print(fit_bivariate)
````

---
## 🅒 Section C — Sup-Wald Joint Significance Test: `sup_wald()`

Summarises the entire (θ, τ) grid into a **single joint test statistic**.  
Based on Troster (2018).

````python
%%R
sw_biv <- sup_wald(
  qq_result = fit_bivariate,  # fitted object from qq_causality() or mqq_causality()
  alpha     = 0.05            # significance level (common: 0.01, 0.05, 0.10)
)

print(sw_biv)
````

---
## 🅓 Section D — Extract Result Matrix: `qq_causality_to_matrix()`

Converts the result object into a plain **numeric matrix** — useful for exporting to CSV or custom plots.

````python
%%R
# Extract the t-statistic matrix
t_matrix <- qq_causality_to_matrix(
  qq_result = fit_bivariate,
  value     = "t_value"  # options: "t_value" (t-statistics) or "slope" (coefficients)
)

cat("── T-Value Matrix (rows = theta/y-quantile, cols = tau/x-quantile) ──\n")
print(round(t_matrix, 3))
````

````python
%%R
# Extract the slope (regression coefficient) matrix
b_matrix <- qq_causality_to_matrix(
  qq_result = fit_bivariate,
  value     = "slope"  # regression slope at each (theta, tau) cell
)

cat("── Slope Matrix ──\n")
print(round(b_matrix, 3))
````

## 🅑 Section B — Multivariate Conditional QQ Causality: `mqq_causality()`

Same as Section A but additionally **conditions on moderators Z** and optional **x × Z interaction terms**.  
Implements the conditional causality test of Sinha et al. (2023).

```r
mqq_causality(x, y, moderators = list(),
              y_quantiles, x_quantiles,
              bandwidth, n_boot, interactions = TRUE,
              cdf_based_kernel, cause_name, effect_name,
              verbose, seed)
```

### B.1 — Simulate Toy Data (Multivariate)

````python
%%R
set.seed(42)
n <- 200

z1 <- rnorm(n)  # moderator 1 (e.g. interest rate)
z2 <- rnorm(n)  # moderator 2 (e.g. exchange rate)
x  <- rnorm(n)  # cause variable

# y depends on lagged x AND lagged z1
y <- 0.3 * c(0, x[-n]) +
     0.2 * c(0, z1[-n]) +
     rnorm(n, sd = 0.4)

cat("Data ready: n =", n, "| Variables: x, y, z1, z2\n")
````

### B.2 — Run `mqq_causality()`

````python
%%R
fit_multivariate <- mqq_causality(
  x          = x,   # REQUIRED ► cause variable
  y          = y,   # REQUIRED ► effect variable

  # REQUIRED ► named list of conditioning / control variables (can be empty list)
  moderators = list(
    Z1 = z1,  # moderator 1 — same length as x and y
    Z2 = z2   # moderator 2 — same length as x and y
  ),

  y_quantiles      = seq(0.1, 0.9, by = 0.1),
  x_quantiles      = seq(0.1, 0.9, by = 0.1),

  bandwidth        = 0.05,
  n_boot           = 200,
  interactions     = TRUE,   # TRUE = include x*Z cross-terms (full conditional test)
  cdf_based_kernel = TRUE,
  cause_name       = "Oil_Price",
  effect_name      = "Stock_Return",
  verbose          = TRUE,
  seed             = 42
)
````

### B.3 — Print Multivariate Results

````python
%%R
print(fit_multivariate)
````

### B.4 — Sup-Wald Test on Multivariate Result

````python
%%R
sw_mult <- sup_wald(fit_multivariate, alpha = 0.05)
print(sw_mult)
````

## ⚙️ Section G — Low-Level Building Blocks (Advanced Users)

### G.1 — Kernel Weights: `qq_weights()`

````python
%%R
w <- qq_weights(
  x         = x,      # numeric vector
  tau       = 0.5,    # quantile level to centre the kernel on (0 to 1)
  h         = 0.05,   # bandwidth
  cdf_based = TRUE    # TRUE = use empirical-CDF distance
)

cat("Number of weights:", length(w), "\n")
cat("Sum of weights:   ", round(sum(w), 4), "(should equal n =", length(x), ")\n")
cat("First 10 weights: ", round(head(w, 10), 4), "\n")
````

### G.2 — Raw Gaussian Kernel: `gaussian_kernel()`

````python
%%R
u <- seq(-3, 3, by = 0.5)  # input values
k <- gaussian_kernel(u)    # Gaussian kernel evaluated at each u

cat("u values: ", u, "\n")
cat("K(u):     ", round(k, 4), "\n")
````

### G.3 — Weighted Quantile Regression: `weighted_qr()`

````python
%%R
# Build design matrix: intercept + x
X_mat <- cbind(1, x)

fit_wqr <- weighted_qr(
  y       = y,      # response vector
  X       = X_mat,  # numeric design matrix (intercept in first column)
  tau     = 0.5,    # quantile level (0 to 1)
  weights = w       # kernel weights from qq_weights()
)

cat("Coefficients (intercept, slope):\n")
print(round(fit_wqr$coef, 6))
````
